# Deterministic research workflow

This local example creates a synthetic price series and computes a trailing return.

```python
import numpy as np
import pandas as pd
from finahinking.features import returns

seed = 20261002
rng = np.random.default_rng(seed)
steps = rng.normal(loc=0.001, scale=0.01, size=30)
prices = pd.Series(100 * (1 + steps).cumprod(), name='synthetic_price')
daily_returns = returns(prices)

assert len(prices) == 30
assert daily_returns.iloc[0] != daily_returns.iloc[0]  # first return is undefined
prices.head()
```

```python
summary = pd.Series({
    'seed': seed,
    'observations': len(prices),
    'cumulative_return': float(prices.iloc[-1] / prices.iloc[0] - 1),
})
summary
```
